## Binary Classification

Here we can be working on binary classficiation for example identifying a cat ..if its a cat...we say 1 else 0. Lets see how an image is represented in computer. To store an image your computer stores three separate matrices corresponding to the red, green, and blue color channels of this image. So if your input image is 64 pixels by 64 pixels, then you would have 3 64 by 64 matrices corresponding to the red, green and blue pixel intensity values for your image. So to turn these pixel intensity values- Into a feature vector, what we're going to do is unrow all of these pixel values into an input feature vector x. We are going to define feature vector corresponding to image which will take all the pixel values from matrices of red,green,blue.<br>
X=\[255 <br>
   231 <br>
   .....so on all red pixels <br>
   255 <br>
   134 <br>
   .....so on all green pixels <br>
   .....so on all blue pixels]<br>
The above pixel values are mentioned in 64 cross 64 matrices of a particular picture. Therefore total dimension of X is 64\*64*3 = 12288 = n. So in binary classification, our goal is to learn a classifier that can input an image represented by this feature vector x. And predict whether the corresponding label y is 1 or 0, that is, whether this is a cat image or a non-cat image. <br>

## Computation graphs

The computations of a neural network are organized in terms of a forward pass or a forward propagation step, in which we compute the output of the neural network, followed by a backward pass or back propagation step, which we use to compute gradients or compute derivatives. The computation graph explains why it is organized this way. <br>
Lets suppose we have some random function i.e J(a,b,c) = 3*(a + b\*c)<br>
Computing this function actually has 3 steps as shown below where we first calculate b*c..then we add a and then multiply the result with 3. And we have made a computation graph below.<br>

## Gradient Descent on m examples

So, what you need to do is really compute derivatives as we showed on the previous training example for each example and average them, and this will give you the overall gradient that you can use to implement the gradient descent. <br>
![regression with multiple examples](reg.png)<br>
So, with all of these calculations, you've just computed the derivatives of the cost function J with respect to each your parameters w_1, w_2 and b. <br>
After these calculations, <br>
w1 = w1 - alpha\*dw1 <br>
w2 = w2 - alpha\*dw2 <br>
b = b - alpha*db <br>
This all is one step of gradient descent, so we have to do these steps multiple times to get our ans.<br>
We have weaknesses where we have to write 2 for loops here. One is over all examples and second is if we have multiple features then we wont write each feature derivative seperately and instead put it in a loop. Hence our algorithms get less efficient and would create a problem if we have huge dataset.<br>
To get rid of loops, we use VECTORIZATION.
